```python
from selenium import webdriver
from selenium.webdriver.common.by import By

def web_scraper(website):

    driver = webdriver.Chrome()
    driver.get(website)

    all_matches = driver.find_element(By.XPATH, '//label[@analytics-event="All matches"]')
    all_matches.click()

    matches = driver.find_elements(By.TAG_NAME, 'tr')

    data = []

    for match in matches:
        data.append(match.text)

    driver.quit()

    return data
```

```python
from openai import OpenAI

OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(
    base_url=OLLAMA_BASE_URL,
    api_key="ollama"
)

def ask_llm(data):

    text_data = "\n".join(data[:20])

    response = ollama.chat.completions.create(
        model="llama3.2",
        messages=[
            {"role": "system", "content": f"Analyze this football match data and summarize it:\n{text_data}"},
            {"role": "user", "content": f"Which team appears most frequently?"}

        ]
    )

    return response.choices[0].message.content
```

```python
website = "https://www.adamchoi.co.uk/overs/detailed"

scraped_data = web_scraper(website)

llm_response = ask_llm(scraped_data)

print(llm_response)
```
